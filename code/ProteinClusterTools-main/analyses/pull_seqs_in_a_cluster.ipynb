```python
cluster162 = pd.read_csv('id_header_clustersat162.csv')
annot = pd.read_csv('annot_table_IPR001223_bact_arch_jackseq_annot_HMMseqs_and_NOhmmexplodeddomains_ANNOT_pfamscan.csv')

cluster162
```

```python
#only take columns Entry and Domain_number from annot
annot = annot[['Entry', 'Total_domains']]
annot_cluster162 = annot.merge(cluster162, left_on='Entry', right_on = 'Uniprot_Accessions',how='left')
annot_cluster162 = annot_cluster162.drop_duplicates(subset=['Entry'], keep='first')
annot_cluster162
```

```python
import numpy as np
import pandas as pd

SEQ_ID = "Uniprot_Accessions"          # assumed: one row == one sequence
DOMCOL = "Total_domains"    # assumed: TOTAL no. of GH18 domains in that sequence
df = annot_cluster162.copy()
d = df.drop_duplicates(subset=SEQ_ID).copy()
d[DOMCOL] = pd.to_numeric(d[DOMCOL], errors="coerce")

# cluster size = no. of sequences per cluster
d["cluster_size"] = d.groupby("cluster")[SEQ_ID].transform("size")

bins   = [0, 1, 2, 4, 9, 19, 49, 99, 499, np.inf]
labels = ["1", "2", "3–4", "5–9", "10–19", "20–49", "50–99", "100–499", "500+"]
d["size_bin"] = pd.cut(d["cluster_size"], bins=bins, labels=labels)

g = d.groupby("size_bin", observed=False)
stats = pd.DataFrame({
    "n_clusters": g["cluster"].nunique(),
    "n_seqs":     g.size(),
    "one_n":      g[DOMCOL].agg(lambda s: (s == 1).sum()),
    "three_n":    g[DOMCOL].agg(lambda s: (s == 3).sum()),
})
stats.loc["Total"] = stats.sum()

stats["one_pct"]   = 100 * stats["one_n"]   / stats["n_seqs"]
stats["three_pct"] = 100 * stats["three_n"] / stats["n_seqs"]

supp8 = pd.DataFrame({
    "Cluster-size bin":   stats.index,
    "No. of clusters":    stats["n_clusters"].map("{:,.0f}".format).values,
    "Total sequences":    stats["n_seqs"].map("{:,.0f}".format).values,
    "One-domain, n (%)":  [f"{n:,.0f} ({p:.1f}%)" for n, p in zip(stats["one_n"], stats["one_pct"])],
    "Three-domain, n (%)":[f"{n:,.0f} ({p:.1f}%)" for n, p in zip(stats["three_n"], stats["three_pct"])],
})

# --- sanity checks: read these before using the table ---
print("rows:", len(df), "| unique", SEQ_ID + ":", df[SEQ_ID].nunique())
print("NaN Domain_number:", d[DOMCOL].isna().sum())
print("Domain_number value counts:\n", d[DOMCOL].value_counts(dropna=False).sort_index())
# does the 'NofM' suffix agree with Domain_number? (i.e. is Domain_number a total, not an index?)
m = d[SEQ_ID].str.extract(r"(\d+)of(\d+)$").astype(float)
print("suffix M == Domain_number:", (m[1] == d[DOMCOL]).mean(),
      "| suffix N == Domain_number:", (m[0] == d[DOMCOL]).mean())

display(supp8)
supp8#.to_csv("supplementary_table_8.csv", index=False)
```
